```python
# Calculate key summary metrics - 25

debt_2000 = global_trend[
    (global_trend["KPI"] == "External Debt Stock") &
    (global_trend["Year"] == 2000)
]["Total Value (US$ Billion)"].iloc[0]

debt_2024 = global_trend[
    (global_trend["KPI"] == "External Debt Stock") &
    (global_trend["Year"] == 2024)
]["Total Value (US$ Billion)"].iloc[0]

absolute_increase = debt_2024 - debt_2000

percentage_increase = (
    absolute_increase / debt_2000 * 100
)

cagr = (
    (debt_2024 / debt_2000) ** (1 / 24) - 1
) * 100

print(f"Debt stock in 2000: US${debt_2000:,.2f} billion")
print(f"Debt stock in 2024: US${debt_2024:,.2f} billion")
print(f"Absolute increase: US${absolute_increase:,.2f} billion")
print(f"Percentage increase: {percentage_increase:.2f}%")
print(f"CAGR, 2000-2024: {cagr:.2f}%")
```
